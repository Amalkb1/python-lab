```python
# a) Square of a number
square = lambda x: x * x
print("Square:", square(5))

# b) Cube of a number
cube = lambda x: x * x * x
print("Cube:", cube(3))

# c) Largest of two numbers
largest = lambda a, b: a if a > b else b
print("Largest:", largest(10, 20))

# d) Check even or odd
check = lambda x: "Even" if x % 2 == 0 else "Odd"
print("Number is:", check(7))

# e) Area of square
area_square = lambda side: side * side
print("Area of Square:", area_square(5))

# Area of rectangle
area_rectangle = lambda length, breadth: length * breadth
print("Area of Rectangle:", area_rectangle(5, 4))

# Area of triangle
area_triangle = lambda base, height: 0.5 * base * height
print("Area of Triangle:", area_triangle(6, 4))
```

# prg 30

```python
class rectangle:
    def __init__(self,length,breadth):
        self.length=length
        self.breadth=breadth
    def area(self):
        return self.length*self.breadth
    def perimeter(self):
        return 2*(self.length+self.breadth)
    
r1=rectangle(10,5)
r2=rectangle(8,6)
print("Rectangle 1 area :",r1.area())
print("Rectangle 1 Perimeter :",r1.perimeter())
print("Rectangle 2 area :",r2.area())
print("Rectangle 2 Perimeter :",r2.perimeter())
if r1.area()>r2.area():
    print("Rectangle 1 has greater area")
elif r1.area()< r2.area():
    print("Rectangle 2 has greater area")
else:
    print("Both rectangle have equal area")
    
```

# prg 31

```python
class Bank:
    def __init__(self, acc_no, name, acc_type, balance):
        self.acc_no = acc_no
        self.name = name
        self.acc_type = acc_type
        self.balance = balance

    def deposit(self, amount):
        if amount > 0:
            self.balance += amount
            print("Deposited:", amount)
        else:
            print("Deposit amount must be positive")

    def withdraw(self, amount):
        if amount <= 0:
            print("Withdrawal amount must be positive")
        elif amount <= self.balance:
            self.balance -= amount
            print("Withdrawn:", amount)
        else:
            print("Insufficient balance")

    def display(self):
        print("Account Number:", self.acc_no)
        print("Name:", self.name)
        print("Account Type:", self.acc_type)
        print("Balance:", self.balance)


b1 = Bank(101, "Amal", "Savings", 5000)
b1.display()
b1.deposit(2000)
b1.withdraw(1000)
print("Final Balance:", b1.balance)
```
